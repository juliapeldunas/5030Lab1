```python
n = 32   #one sample

data = np.random.uniform(-np.sqrt(3), np.sqrt(3), n)

mean = np.mean(data)
mean2 = mean * np.sqrt(n)

mean2
```

```python
results = []

for i in range(5000): #repeat 5000 times creating the 32 new variables, fidning their mean, and multiplying by sqrt(32).
    data = np.random.uniform(-np.sqrt(3), np.sqrt(3), n)
    mean = np.mean(data)
    mean2 = mean * np.sqrt(n)
    results.append(mean2)
```

```python
from statsmodels.distributions.empirical_distribution import ECDF
import matplotlib.pyplot as plt

ecdf = ECDF(results)

plt.plot(ecdf.x, ecdf.y)
plt.xlabel("Standardized Sample Mean")
plt.ylabel("ECDF")
plt.show()
```

This approaches a normal distribtuion because the ECDF approaches 1 as the sample size n increases, showing that the standardized sample means become more normally distributed as n gets larger.
