```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

rng = np.random.default_rng(20260717)
```

```python
# suppose there are k=100000 cities, ranked from largest to smallest
# create the ranks 

K = 10000

ranks = np.linspace(1, K, num=K)
```

```python
# Zipf's law says that the size of the city with rank r is proportional to 
# x_r = (K/r)^s. Start with s =1 

s = 1

def Zipf(ranks) :
    return (K / ranks) ** s
```

```python
# randomly sample n = 5000 ranks, with each rank equally likely,
# and record the corresponding size

n = 5000

outputs =[]

random_samples = np.random.choice(ranks, size = n)

random_samples[4]

for i in range(n) :
    result = Zipf(random_samples[i])
    outputs.append(result)


outputs
```

```python
# plot the ECDF of the simulated sizes

plt.ecdf(outputs)
```

What distribution does this generate as K becomes large?

Exponential 

How does the distribution change as you adjust s?

As s increases, the curve becomes a lot steeper, approaching an entirely vertical asymptote at x = 0 and a horizontal line from x = 0 to x = 1
