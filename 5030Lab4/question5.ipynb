```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy import stats

rng = np.random.default_rng(20260717)
```

```python
# draw K = 20 values from uniform(0,1)
K = 20

results = rng.uniform(0, 1, K)
print(results)
```

```python
# sort them by size from smallest to largest

results.sort()

results
```

```python
# record the largest value 

largest = results[K-1]

largest
```

```python
# repeat n times and plot the ECDF 

n = 5000

outputs = []

for i in range(n):
    results = rng.uniform(0, 1, K)
    results.sort()
    largest = results[K-1]
    outputs.append(largest)
  
plt.ecdf(outputs)
# plt.xlim(.9, 1)
```

```python
# repeat the previous steps for each rank from 1 to K

n = 20

for j in range(K):
    outputs = []
    for i in range(n):
        results = rng.uniform(0, 1, K)
        results.sort()
        rank = results[j]
        outputs.append(rank)
    plt.ecdf(outputs)
    plt.xlabel(f"rank = {j}")
    plt.xlim(0, 1)
    plt.show()
```

What distribution(s) does this process generate? How does the distribution depend on the rank?

This is the beta distribution. As the rank increases, the curve moves rightwards across the graph, towards x = 1.
